```python
# 1. Smoothening and Sharpening
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(r'D:\SEM_5\DSIP\DSIP_practical\EXP_09\cameraman.jpg')

if image is None:
    print("Error: Image not found. Check the file path.")
else:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    kernel_size = (5, 5)
    sigma = 1.5

    gaussian_kernel_1D = cv2.getGaussianKernel(5, sigma)
    gaussian_kernel = np.outer(gaussian_kernel_1D, gaussian_kernel_1D)

    smoothed_image = cv2.filter2D(image, -1, gaussian_kernel)
    smoothed_rgb = cv2.cvtColor(smoothed_image, cv2.COLOR_BGR2RGB)

    sharpening_kernel = np.array([
        [-1, -1, -1],
        [-1,  9, -1],
        [-1, -1, -1]
    ], dtype=np.float32)

    sharpened_image = cv2.filter2D(image, -1, sharpening_kernel)
    sharpened_rgb = cv2.cvtColor(sharpened_image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(smoothed_rgb)
    plt.title("Smoothed Image")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(sharpened_rgb)
    plt.title("Sharpened Image")
    plt.axis("off")

    plt.show()
```

```python
# 2. Smoothening using Averaging Linear Filter

import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(r'D:\SEM_5\DSIP\DSIP_practical\EXP_09\cameraman.jpg')

if image is None:
    print("Error: Image not found. Check the file path.")
else:
    kernel_size = (5, 5)

    kernel = np.ones(
        kernel_size,
        dtype=np.float32
    ) / (kernel_size[0] * kernel_size[1])

    smoothed_image = cv2.filter2D(image, -1, kernel)

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    smoothed_rgb = cv2.cvtColor(smoothed_image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(smoothed_rgb)
    plt.title("Averaging Filter")
    plt.axis("off")

    plt.show()
```

```python
# 3. Smoothening using Median Filter

import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r'D:\SEM_5\DSIP\DSIP_practical\EXP_09\cameraman.jpg')

if image is None:
    print("Error: Image not found. Check the file path.")
else:
    kernel_size = 5

    smoothed_image = cv2.medianBlur(image, kernel_size)

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    smoothed_rgb = cv2.cvtColor(smoothed_image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(smoothed_rgb)
    plt.title("Median Filter Smoothed Image")
    plt.axis("off")

    plt.show()
```

```python
# 4. Sharpening using Spatial High-Pass Filter

import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(r'D:\SEM_5\DSIP\DSIP_practical\EXP_09\cameraman.jpg')

if image is None:
    print("Error: Image not found. Check the file path.")
else:
    blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

    laplacian_kernel = np.array([
        [0, -1, 0],
        [-1, 5, -1],
        [0, -1, 0]
    ], dtype=np.float32)

    sharpened_image = cv2.filter2D(
        blurred_image,
        -1,
        laplacian_kernel
    )

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    blurred_rgb = cv2.cvtColor(blurred_image, cv2.COLOR_BGR2RGB)
    sharpened_rgb = cv2.cvtColor(sharpened_image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(blurred_rgb)
    plt.title("Blurred Image")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(sharpened_rgb)
    plt.title("Sharpened Image")
    plt.axis("off")

    plt.show()
```
