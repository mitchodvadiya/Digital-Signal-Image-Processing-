```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("cameraman.jpg", cv2.IMREAD_GRAYSCALE)
if image is None:
    raise FileNotFoundError("Could not load 'cameraman.jpg' from the notebook directory.")

_, binary_image = cv2.threshold(image, 127, 1, cv2.THRESH_BINARY)
pattern = np.array([[-1, 1, -1], [1, 1, 1], [-1, 1, -1]], dtype=np.int8)
result = cv2.morphologyEx(binary_image, cv2.MORPH_HITMISS, pattern)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(image, cmap="gray")
axes[0].set_title("Original")
axes[1].imshow(result, cmap="gray")
axes[1].set_title("Hit-or-miss")
for ax in axes:
    ax.axis("off")  
plt.tight_layout()
plt.show()
```
